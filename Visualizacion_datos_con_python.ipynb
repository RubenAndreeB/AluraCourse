# Visualizacion de datos con Python: Matplotlib, Seaborn y Plotly

## 1. Conociendo la biblioteca Matplotlib

### Conociendo los datos

```python
import pandas as pd
```

```python
ult_informacion = 'https://raw.githubusercontent.com/alura-es-cursos/2063-visualizacion-de-datos-con-bibliotecas-de-python/refs/heads/main/inmigrantes_canada.csv'
df = pd.read_csv(ult_informacion)
df.head()
```

```python
df.shape
```

```python
df.info()
```

```python
df.set_index('Pais',inplace=True)
df
```

```python
anos = list(map(str, range(1980,2014)))
anos
```

```python
colombia = df.loc['Colombia',anos]
colombia
```

```python
col_dict = {'Año': colombia.index.tolist(),
            'Inmigrantes': colombia.values.tolist()}
col_dict
```

```python
datos_col = pd.DataFrame(col_dict)
datos_col.tail()
```

### Creando el primer grafico con matplotlib

```python
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize =(8,4))
plt.plot(datos_col['Año'], datos_col['Inmigrantes'])
plt.xticks(list(map(str, range(1980, 2014, 5))))
plt.show()
```

### Añadiendo títulos y rótulos

```python
plt.figure(figsize =(8,4))
plt.plot(datos_col['Año'], datos_col['Inmigrantes'])
plt.xticks(list(map(str, range(1980, 2014, 5))))
plt.title('Inmigracion de colombianos a Canada')
plt.xlabel('Año')
plt.ylabel('Numero de Inmigrantes')
plt.grid(True)
plt.show()
```

## 2. Creando figuras con Matplotlib

### Creando una figura

```python
fig, ax = plt.subplots(figsize=(8,4))
ax.plot(datos_col['Año'], datos_col['Inmigrantes'])
ax.xaxis.set_major_locator(plt.MultipleLocator(5))
plt.show()
```

### Añdiendo títulos y rótulos en la figura

```python
fig, ax = plt.subplots(figsize=(8,4))
ax.plot(datos_col['Año'], datos_col['Inmigrantes'])
ax.xaxis.set_major_locator(plt.MultipleLocator(3))
ax.set_title('Inmigracion de colombianos a Canada\n1980 - 2013')
ax.set_xlabel('Año')
ax.set_ylabel('Numero de Inmigrantes')
plt.show()
```

### Creando subplots en uns dirección

```python
fig, axs = plt.subplots(1, 2, figsize=(10,4))
axs[0].plot(datos_col['Año'], datos_col['Inmigrantes'])
axs[0].xaxis.set_major_locator(plt.MultipleLocator(5))
axs[0].set_title('Inmigracion de colombianos hacia Canada\n1980 - 2013')
axs[0].set_xlabel('Año')
axs[0].set_ylabel('Numero de Inmigrantes')
axs[0].grid()

axs[1].boxplot(datos_col['Inmigrantes'])
axs[1].set_title('Boxplot de la inmigracion de colombianos\nhacia Canada 1980 - 2013')
axs[1].set_xlabel('Colombia')
axs[1].set_ylabel('Numero de Inmigrantes')
axs[1].grid()
plt.show()
```

### Creando subplots en dos direcciones #1

```python
fig, axs = plt.subplots(2, 2, figsize=(10,6))
fig.subplots_adjust(hspace=0.5, wspace=0.3)

axs[0,0].plot(df.loc['Colombia',anos])
axs[0,0].set_title('Colombia')

axs[0,1].plot(df.loc['Brasil',anos])
axs[0,1].set_title('Brasil')

axs[1,0].plot(df.loc['Argentina',anos])
axs[1,0].set_title('Argentina')

axs[1,1].plot(df.loc['Perú',anos])
axs[1,1].set_title('Perú')

for ax in axs.flat:
    ax.xaxis.set_major_locator(plt.MultipleLocator(5))
    ax.set_xlabel('Año')
    ax.set_ylabel('Numero de Inmigrantes')
    ax.grid()

plt.show()
```

### Creando subplots en dos direcciones #2

```python
fig, axs = plt.subplots(2, 2, figsize=(10,6))
fig.subplots_adjust(hspace=0.5, wspace=0.3)
fig.suptitle('Inmigracion de los 4 mayores paises SudAmericanos\nhacia Canada desde 1980 a 2013')
axs[0,0].plot(df.loc['Colombia',anos])
axs[0,0].set_title('Colombia')

axs[0,1].plot(df.loc['Brasil',anos])
axs[0,1].set_title('Brasil')

axs[1,0].plot(df.loc['Argentina',anos])
axs[1,0].set_title('Argentina')

axs[1,1].plot(df.loc['Perú',anos])
axs[1,1].set_title('Perú')

ymin = 0
ymax = 7000

for ax in axs.ravel():
    ax.set_ylim(ymin, ymax)

for ax in axs.flat:
    ax.xaxis.set_major_locator(plt.MultipleLocator(5))
    ax.set_xlabel('Año')
    ax.set_ylabel('Numero de Inmigrantes')
    ax.grid()

plt.show()
```

## 3. Personalizando graficos con Matplotlib

### Personalizando visualizaciones con matplotlib

```python
fig, ax = plt.subplots(figsize=(8,4))
ax.plot(datos_col['Año'], datos_col['Inmigrantes'], lw=3, marker = '.')
ax.xaxis.set_major_locator(plt.MultipleLocator(5))
ax.set_title('Inmigracion de colombianos a Canada\n1980 - 2013', fontsize=18, loc='left')
ax.set_xlabel('Año', fontsize=14)
ax.set_ylabel('Numero de Inmigrantes', fontsize=14)
ax.yaxis.set_tick_params(labelsize=12)
ax.xaxis.set_tick_params(labelsize=12)
ax.yaxis.set_tick_params(labelsize=12)
ax.grid()
plt.show()
```

### Alterando colores #1

```python
fig, ax = plt.subplots(figsize=(8,3))
ax.plot(datos_col['Año'], datos_col['Inmigrantes'], lw=2, marker = '.', color = 'green')
ax.xaxis.set_major_locator(plt.MultipleLocator(5))
ax.set_title('Inmigracion de colombianos a Canada\n1980 - 2013', fontsize=18, loc='left')
ax.set_xlabel('Año', fontsize=14)
ax.set_ylabel('Numero de Inmigrantes', fontsize=14)
ax.yaxis.set_tick_params(labelsize=12)
ax.xaxis.set_tick_params(labelsize=12)
ax.yaxis.set_tick_params(labelsize=12)
ax.grid(linestyle='--')
ax.set_ylim(0,7000)
plt.show()
```

```python
sudamerica = df.query('Region == "América del Sur"')
sudamerica.head()
```

```python
fig, ax = plt.subplots(figsize = (12,6))
ax.barh(sudamerica.index, sudamerica['Total'], color=colores)
ax.set_title('Inmigracion de sudamericanos hacia Canda\ndurante el periodo de 1980 - 2013',
loc = 'left', fontsize = 18)
ax.set_ylabel(' ')
ax.set_xlabel('Numero de Inmigrantes', fontsize = 14)
ax.xaxis.set_tick_params(labelsize = 12)
ax.yaxis.set_tick_params(labelsize = 12)
plt.show()
```

```python
sudamerica_sorted = sudamerica.sort_values('Total', ascending=True)
fig, ax = plt.subplots(figsize = (12,6))
ax.barh(sudamerica_sorted.index, sudamerica_sorted['Total'], color=colores)
ax.set_title('Inmigracion de sudamericanos hacia Canda\ndurante el periodo de 1980 - 2013',
loc = 'left', fontsize = 18)
ax.set_ylabel(' ')
ax.set_xlabel('Numero de Inmigrantes', fontsize = 14)
ax.xaxis.set_tick_params(labelsize = 12)
ax.yaxis.set_tick_params(labelsize = 12)
plt.show()
```

### Utilizando diversas palettes

```python
def generar_grafico(palette):
  fig, ax = plt.subplots(figsize=(8,4))
  ax = sns.barplot(data = top_10, x='Total', hue=top_10.index, orient='h',
                   palette = palette)
  ax.set_title("Top 10 paises con mayor migracion hacia Canada\ndurante el periodo de 1980 - 2013",
              loc = 'left', fontsize = 16)
  ax.set_xlabel('Numero de Inmigrantes', fontsize = 14)
  ax.set_ylabel(' ')
  plt.show()
```

## Graficos interactivos con Plotly
> Documentacion: https://plotly.com/python/

### Creando nuestro primer gráfico interactivo

```python
import plotly.express as px
```

```python
fig = px.line(datos_col, x='Año', y='Inmigrantes')
fig.show()
```

```python
fig = px.line(datos_col, x='Año', y='Inmigrantes')
fig.update_layout(width = 600, height = 400,
                  xaxis={'tickangle': -45})
fig.show()
```

### Añadiendo rótulos y modificando fuentes

```python
fig = px.line(datos_col, x='Año', y='Inmigrantes', title='Inmigracion de colombianos hacia Canada\n1980 - 2013')
fig.update_layout(width = 800, height = 400, xaxis={'tickangle': -45},
                  xaxis_title = 'Año', yaxis_title = 'Numero de Inmigrantes',
                  font_family='Arial', font_size=14, font_color='grey')
fig.show()
```

### Alterando colores con Plotly

```python
fig = px.line(datos_col, x='Año', y='Inmigrantes', title='Inmigracion de colombianos hacia Canada\n1980 - 2013')
fig.update_traces(line_color='green', line_width=3)
fig.update_layout(width = 800, height = 400, xaxis={'tickangle': -45},
                  xaxis_title = 'Año', yaxis_title = 'Numero de Inmigrantes',
                  font_family='Arial', font_size=14, font_color='grey')
fig.show()
```
