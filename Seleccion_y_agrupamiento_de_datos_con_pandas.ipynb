# Agrupamiento de los datos
---

## Remodelando el DataFrame

Vamos a modifiar el formato para que las columnas referentes a cada anio pases a ser una unica columna.

```python
emision_gases.loc[:,'Nível 1 - Setor':'Produto'].columns
```

```python
columnas_info = list(emision_gases.loc[:,'Nível 1 - Setor':'Produto'].columns)
```

```python
emision_gases.loc[:,1970:2021].columns
```

```python
columnas_emision = list(emision_gases.loc[:,1970:2021].columns)
```

Para realizarla transformacion del dataframe utilizamos el metodo melt(). Este metodo tiene como objetivo transformar un DataFramede un formato ancho a largo.

```python
emisiones_por_ano = emision_gases.melt(id_vars = columnas_info, value_vars = columnas_emision, var_name = 'Año', value_name = 'Emision')
```

## Agrupando los gases

Vamos a realizar un analisis de la cantidad de emision totoal por tipo de gas. Usando `groupby()`

```python
emisiones_por_ano.groupby('Gás')
```

Podemos revisr cuales fueron agrupados en cada grupo con el uso del atributo `groups` y realizar selecciones de grupos especificos con el metodo `get_group()`

```python
emisiones_por_ano.groupby('Gás').groups
```

```python
emisiones_por_ano.groupby('Gás').get_group('CO2 (t)')
```

Los metodos de agregacion psoibles con la utilizacion de group_by() son:
- media(): calcula la media de los grupos
- suma(): calcula la suma de los valores del grupo
- size() : calcula el tamaño del grupo
- count(): calcula el conteo del grupo
- std(): desviación estándar de grupos
- var(): calcula la variación de grupos
- sem(): error estándar de la media del grupo
- describe(): genera estadísticas descriptivas
- first() : calcula el primero de los valores del grupo
- last(): calcula el último de los valores del grupo
- nth(): obtiene el enésimo valor del grupo
- min(): calcula el número mínimo de valores en el grupo
- max(): calcula el número máximo de valores en el grupo

```python
emisiones_por_ano.groupby('Gás').sum(numeric_only = True)
```

```python
emisiones_por_gas = emisiones_por_ano.groupby('Gás').sum(numeric_only = True).sort_values('Emision', ascending = False)
emisiones_por_gas
```

```python
emisiones_por_gas.plot(kind = 'barh', figsize = (10,6));
```

```python
emisiones_por_gas.iloc[0:9]
```

```python
(emisiones_por_gas.iloc[0:9].sum() / emisiones_por_gas.sum()).iloc[0] * 100
```

```python
print(f'Las emisiones de CO2 corresponden a {(emisiones_por_gas.iloc[0:9].sum() / emisiones_por_gas.sum()).iloc[0] * 100:.2f} % de las emiisones de gases totales de 1970 a 2021')
```

# Agrupamientos Multi-Index

## Emision de gases por sector
Ahora necesitamos encontrar dos informaciones
- Cuales son los sectores mas contaminantes por cada tipo de gas
- Cuales son los gases mas contaminantes por cada sector

```python
gas_por_sector = emisiones_por_ano.groupby(['Gás', 'Nível 1 - Setor']).sum(numeric_only = True)
```

```python
gas_por_sector
```

Para descubrir el sector que emite mayor cantidad de CO2, podemos realizar una selccion especifica de gas C02 mediante el metodo cross section `xc()`, extrayendo el valor maximo y el indiice maximo con los metodos `max()` y `idxmax()`.

```python
gas_por_sector.xs('CO2 (t)', level = 0)
```

```python
gas_por_sector.xs(('CO2 (t)', 'Mudança de Uso da Terra e Floresta'), level = [0,1])
```

```python
gas_por_sector.xs('CO2 (t)', level = 0).max()
```

```python
gas_por_sector.xs('CO2 (t)', level = 0).idxmax()
```

de manera automatica:

```python
gas_por_sector.groupby(level = 0).idxmax()
```

```python
gas_por_sector.groupby(level = 0).max()
```

```python
valores_max = gas_por_sector.groupby(level = 0).idxmax().values
```

```python
max_emission_values = gas_por_sector.groupby(level=0)['Emision'].max()
max_emission_sectors_info = gas_por_sector.groupby(level=0)['Emision'].idxmax()
tabla_totalizada = pd.DataFrame({'Gas': max_emission_values.index, 'Emision': [info[1] for info in max_emission_sectors_info.values], 'Cantidad de emision': max_emission_values.values})
tabla_totalizada
```
